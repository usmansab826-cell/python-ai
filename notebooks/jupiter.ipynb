```python
email = "Rizwan.sir@yahoo.pk"

username = email.split("@")[0]
domain = email.split("@")[1].split(".")[0]
extension = email.split(".")[-1]

print("Username:", username)
print("Domain:", domain)
print("Extension:", extension)
```

```python

from datetime import datetime

email = "usman@gmail.com"

username = email.split("@")[0]
domain = email.split("@")[1].split(".")[0]
extension = email.split(".")[-1]

now = datetime.now()

print("Username:", username)
print("Domain:", domain)
print("Extension:", extension)
print("Date:", now.strftime("%d-%m-%Y"))
print("Time:", now.strftime("%I:%M %p"))
```

```python
email = input("Enter an email address: ").strip()
parts = email.split("@")

if len(parts) != 2 or not parts[0] or "." not in parts[1]:
    print("Please enter an email address in the format name@example.com.")
else:
    domain, extension = parts[1].rsplit(".", maxsplit=1)
    if not domain or not extension:
        print("Please enter an email address in the format name@example.com.")
    else:
        print("Username:", parts[0])
        print("Domain:", domain)
        print("Extension:", extension)
```
